```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
df["Day_of_Week"] = df["Date"].dt.dayofweek
```

```python
df["Revenue_Per_Passenger"] = (
    df["Revenue"] / df["Passengers"]
)
```

```python
conditions = [
    df["Delay_Minutes"] == 0,
    df["Delay_Minutes"].between(1, 15),
    df["Delay_Minutes"].between(16, 60),
    df["Delay_Minutes"] > 60
]

categories = [
    "On Time",
    "Slight Delay",
    "Moderate Delay",
    "Heavy Delay"
]

df["Delay_Category"] = np.select(
    conditions,
    categories,
    default="Unknown"
)
```

```python
df["Delay_Category"].value_counts()
```

```python
conditions = [
    df["Distance_KM"] <= 300,
    df["Distance_KM"].between(301, 700),
    df["Distance_KM"].between(701, 1200),
    df["Distance_KM"] > 1200
]

categories = [
    "Short",
    "Medium",
    "Long",
    "Very Long"
]

df["Distance_Category"] = np.select(
    conditions,
    categories,
    default="Unknown"
)
```

**Do cancelled journeys have different delay patterns from non-cancelled journeys?**

```python
delay_by_cancellation = (
    df.groupby("Cancelled")["Delay_Minutes"]
      .agg(["count", "mean", "median", "min", "max"])
)

print(delay_by_cancellation)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Cancelled",
    y="Delay_Minutes"
)

plt.title("Delay Distribution by Cancellation Status")
plt.xlabel("Cancelled")
plt.ylabel("Delay (Minutes)")

plt.tight_layout()
plt.show()
```

```python
avg_delay_cancel = df.groupby("Cancelled")["Delay_Minutes"].mean()

print(avg_delay_cancel)

plt.figure(figsize=(7, 5))

avg_delay_cancel.plot(kind="bar")

plt.title("Average Delay by Cancellation Status")
plt.xlabel("Cancelled")
plt.ylabel("Average Delay (Minutes)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
delay_distance = (
    df.groupby("Distance_Category", observed=False)["Delay_Minutes"]
      .agg(["count", "mean", "median", "min", "max"])
)

print(delay_distance)
```
